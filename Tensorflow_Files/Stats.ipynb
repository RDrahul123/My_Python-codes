```python
import statistics
import numpy as np
import pandas as pd
```

```python
# Sample data
data = [12, 15, 18, 22, 25, 25, 30, 35, 40, 45]
```

```python
# Using built-in statistics module
print("Mean:", statistics.mean(data))
print("Median:", statistics.median(data))
print("Mode:", statistics.mode(data))
print("Variance:", statistics.variance(data))       # sample variance
print("Standard Deviation:", statistics.stdev(data)) # sample std dev
```

```python
# Using NumPy
arr = np.array(data)

print("\nNumPy Mean:", np.mean(arr))
print("NumPy Median:", np.median(arr))
print("NumPy Std Dev:", np.std(arr))         # population std dev by default
print("NumPy Std Dev (sample):", np.std(arr, ddof=1))
print("NumPy Variance:", np.var(arr))
print("Min:", np.min(arr))
print("Max:", np.max(arr))
print("Range:", np.max(arr) - np.min(arr))
```

```python
# Percentiles / Quartiles
print("25th percentile (Q1):", np.percentile(arr, 25))
print("50th percentile (Median):", np.percentile(arr, 50))
print("75th percentile (Q3):", np.percentile(arr, 75))
```

```python
# Interquartile Range (IQR)
Q1 = np.percentile(arr, 25)
Q3 = np.percentile(arr, 75)

IQR = Q3 - Q1

print("IQR:", IQR)
```

```python
minimum = np.min(arr)
Q1 = np.percentile(arr, 25)
median = np.median(arr)
Q3 = np.percentile(arr, 75)
maximum = np.max(arr)

print("Minimum:", minimum)
print("Q1:", Q1)
print("Median:", median)
print("Q3:", Q3)
print("Maximum:", maximum)
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([12, 15, 18, 20, 22, 25, 28, 30, 35, 40])

mean = np.mean(data)
std = np.std(data, ddof=1)

plt.scatter(data, np.zeros(len(data)), s=80)

plt.axvline(mean, linestyle="--", label=f"Mean = {mean:.2f}")

plt.axvline(
    mean + std,
    linestyle=":",
    label=f"Mean + STD = {mean + std:.2f}"
)

plt.axvline(
    mean - std,
    linestyle=":",
    label=f"Mean - STD = {mean - std:.2f}"
)

plt.yticks([])
plt.xlabel("Values")
plt.title("Standard Deviation Around the Mean")
plt.legend()

plt.show()
```
